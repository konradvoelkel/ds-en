# Introduction to Python

## Where Python sits

Python has a different design philosophy from C, C++, Perl, Lisp or Pascal, but takes many influences from earlier languages. An active community process keeps the language evolving.

Python is designed to be particularly readable, with a syntax that lets programmers express concepts in fewer lines of code. Clearly, Python is a high-level programming language, further away from concrete machine models than C.
Like Java, Python uses garbage collection, so you do not have to reserve and release memory explicitly in order to hold values in variables.

Python lets you use functional programming paradigms as in R or Haskell, but also object-oriented ones as in C# or Java.
Unlike those languages, Python is *dynamically* typed: the type belongs to the value, not to the name. A variable name can be assigned a string, then immediately afterwards a boolean, and then an integer. Anyone who prefers stricter types can by now work with type annotations and special linters (`mypy`, say); the interpreter itself does not check them.

There are two classical ways of using Python.
With a REPL (read-eval-print loop), similar to a shell on the command line, you can quickly check whether short snippets of code are syntactically correct and behave as you expect.
Every user of Python should get familiar with the interpreter, that is, with the REPL.

By compiling a program (.py) to Python bytecode (.pyc) and having Python execute the bytecode, more complex tasks can be solved. From one file, others can be imported as modules.

Originally, packages (which is what libraries are called in Python) were added to a Python environment with the program pip (Pip Installs Packages). It draws on the PyPI (Python Package Index), say (the analogue of CPAN for Perl, CRAN for R, CTAN for TeX). Only once a package has been downloaded and installed can you import it into your code and use it.

So that the packages of different projects do not get in each other's way, they are not installed system-wide but into a *virtual environment* (venv) per project. The tool [uv](https://docs.astral.sh/uv/) does both in one step: it reads the required packages from a project's `pyproject.toml`, creates the environment in the subdirectory `.venv` and downloads the packages from PyPI. For these notes, `uv sync` in the project directory is enough; after that `uv run python` starts the interpreter with all the packages, and `uv run jupyter lab` the notebook interface. If you prefer to work without uv, you get the same with `python -m venv .venv`, activating the environment, and `pip install -r requirements.txt`. The Anaconda distribution with its conda package manager, once widespread in data science, works too, but is not used in these notes.

A (Jupyter/IPython) notebook is a document consisting of several cells. Each cell can contain either text, mathematics (LaTeX), or Python code, which is then executed directly (like a browser-based REPL).
These notebooks (.ipynb) are implemented as JSON documents.

This book is also a collection of notebooks, which Jupyter Book compiles into a single coherent set of lecture notes.

### Four things to remember

**1. The REPL is our friend.** Python is an interpreted language, and the interpreter can be used interactively: `python` in a console starts the REPL (read-eval-print loop), where you type in expressions and see the result at once. Every Jupyter cell is essentially the same thing. If you do not know what a piece of code does, run it instead of thinking about it – and `help(obj)` and `dir(obj)` will tell you what an object can do.

**2. Indentation instead of braces.** Blocks in Python are not delimited by curly braces but by indentation. This enforces readable code and punishes sloppiness immediately with an `IndentationError`.

**3. Everything is an object.** Numbers too, functions too, modules too. An integer has methods, and a function can be assigned to a variable or passed as an argument – every `apply` and every `key=` argument later on rests on that.

**4. Language models know Python.** Of all programming languages, Python is the one with the most public code available, and language models are correspondingly good at it. That is a real help and a real trap at the same time: generated code looks plausible even when it is wrong, and it does not become correct by running. Check it as you would check somebody else's code – and stick to the disclosure rule for AI tools on the [landing page](landing-page.md).

```python
print((3).bit_length())      # even a number has methods
print(type(print))           # even a function is an object

square = lambda x: x**2      # functions can be assigned to variables
print(list(map(square, [1, 2, 3])))
```

## The essentials

To understand the documentation and the code of Python modules quickly, you have to know Python's syntactic quirks and concepts. Some of the most important ones are collected here.

### Data types

**Numbers** are represented either as integers (int) or as floating point numbers (float), and there are even complex numbers (complex).

Three things are different here from what you know from C or Java. First, Python integers have **no fixed width and do not overflow**: $2^{100}$ is a perfectly ordinary number, and Python enlarges the internal representation as needed. Second, `float`s are floating point numbers following **IEEE 754 binary64** (what C calls `double`) – with everything that entails, in particular that most decimal fractions are not exactly representable. Third, Python is strictly but **dynamically** typed: the type belongs to the value, not to the variable name, and the same name may carry different types one after another.

```python
print("23 is", type(23), "and 23.0 is", type(23.0))
print("and 23+2j is", type(23+2j))
print("adding an int and a float gives", type(23+23.0))
print("dividing integers gives a float, so 5/2 =", 5/2)
print("if we do not want that, we can compute 5//2 =", 5//2)

print("no overflow:", 2**100)
print("IEEE 754:", 0.1 + 0.2 == 0.3, "since 0.1+0.2 =", 0.1 + 0.2)
print("modulo takes the sign of the divisor:", 7 % 3, -7 % 3, "which goes with", -7 // 3)
```

The middle line is the reason why floating point numbers are **never tested for equality**. Instead of `a == b` one writes `abs(a - b) < 1e-9` or, in Numpy, `np.isclose(a, b)` – this will meet us throughout the book. The last line shows a quirk that matters when computing with indices: in Python, `%` always takes the sign of the *divisor*, so `-7 % 3` is $2$ and not $-1$ as in C or Java, and `//` rounds down to match, rather than towards zero.

**Strings** are Unicode throughout, falling back on UTF-8 when nothing else is specified. These days that usually means in practice that source files cope with umlauts and accented characters, but that for relatively old files (from the Python 2 era in particular) the encoding may have to be changed or specified. It will not concern us.

Quotation marks may be single or double, `'Data'` and `"Data"` are the same thing – handy when quotation marks occur inside the string itself. Three quotation marks enclose multi-line strings. Something else matters more: strings are **immutable**. A single character cannot be overwritten, and every method such as `upper()` returns a *new* string instead of changing the old one. In exchange, the same slicing works on strings as on lists.


```python
s = "Data Science"
print(s[0], "|", s[5:], "|", s[-7:])   # slicing works on strings too

try:
    s[0] = "d"
except TypeError as e:
    print("does not work:", e)

print(s.upper(), "| the original is unchanged:", s)
```

**Truth values and `None`.** The type `bool` has the two values `True` and `False`, and it is a special case of `int` – `True == 1` is true. More interesting is that Python can read any value in a condition as a truth value (*truthiness*): the number $0$, the empty string, the empty list, the empty dictionary and `None` count as false, everything else as true. That is why one writes `if items:` rather than `if len(items) > 0:`.

`None` is the only value of type `NoneType` and plays the role that `null` has in Java: "nothing here". A function without a `return` returns `None`. You test for `None` with `is None`, not with `== None` – `is` compares identity, and there is only ever one `None` in the whole program. In data, `None` meets us as a sibling of `NaN` in the chapter [Data cleaning](data-cleaning.ipynb).

```python
print(bool(0), bool(""), bool([]), bool([0]), bool(None))   # what counts as false?
print("bool is an int:", True == 1, True + True)

x = None
if x is None:                      # not "x == None"
    print("x holds nothing")

def without_return():
    pass

print("return value without return:", without_return())
```

**Dictionaries** are maps with a finite domain; the mathematical statement $f \colon \{a,b\} \to \mathbb{R},\ a \mapsto 1,\ b \mapsto 2$ with $f(a) = 1$, for instance, is written as

```python
f = {"a":1, "b":2}
if f == dict(a=1, b=2):
    print("Syntactic sugar occasionally lets us do without the quotation marks.")
if f["a"] == 1:
    print("All as planned")
print(f, "is a", type(f))
try:
    print(f["c"])
    print("the interpreter never gets to this point")
except KeyError:
    f["c"] = "You have to watch where the square and where the round brackets go"
finally:
    print(f["c"]) # but now it does!

try:
    {[1, 2]: "does not work"}       # a list is mutable
except TypeError as e:
    print("not a valid key:", e)

print({(1, 2): "tuples do work"}[(1, 2)])
```

Under the hood a dictionary is a **hash map**: the key is computed into a number, and that number points to the place where the value sits. This is why access is fast independently of the size of the dictionary – and why only **immutable** objects may be keys. A key that changed afterwards would be lying in the wrong place. Numbers, strings and tuples are fine, then; lists and dictionaries are not. That is exactly what the last two lines above show.

**Tuples** are immutable lists of fixed length, so pairs or triples, for instance. Careful: addition is not element-wise but list concatenation, producing a new tuple (different from the old one):

```python
pair = (2,3)
print(pair, "is a", type(pair), "of length", len(pair))
triple = (4,5,6)
print(pair, "+", triple, "=", pair+triple)
print((0,1), "* 3 =", (0,1)*3)
```

**Sets** are unordered lists, and they too are written with curly braces:

```python
a_set = {3,2,1}
print(a_set)
another_set = set((2,3,1))
a_set == another_set
```

**Lists** are ordered and can be extended. Under the hood a list in CPython is an array of pointers, replaced by a larger one when needed. As a user you notice nothing of this, and it is usually irrelevant to questions of performance. A list comprehension is a way of iterating over a list in order to build another one.

```python
items = [3,1,4,1,5,9]
if 5 in items:
    print(items, "of type", type(items), "has length", len(items))
print(items + [-1,-1,-1])
items += [0]
print(items)
triple = (1,2,3)
print(list(triple))
# here comes the list comprehension:
shifted = [x+2 for x in items]
print(shifted)
```

**Slicing** is what it is called when you select part of a list in Python with the colon syntax.

```python
numbers = range(0,10)  # careful, this starts at 0 and 10 is not included.
print(numbers, type(numbers))
numbers = list(numbers)
print(numbers, type(numbers))
print(numbers[1:9]) # here the 9 is not included either.
print(numbers[:5] + numbers[5:])
print("negative indices work too:", numbers[-1:])
print(numbers[:-1] + numbers[-1:])
```

**Generators** are objects that generalise lists in one respect: you can ask them for the next element. But no whole list has to be held in memory (nor computed in advance), which is why generator expressions are the right choice in many places for data-processing pipelines inside Python.
Every list, every tuple and every `range` object can be iterated over in just the same way; `range` is not a generator but a "lazy" sequence, which likewise produces its elements only on demand.

```python
gen = (x + 2 for x in range(0,5))
print(gen, "has type", type(gen))
# and only now is the computation carried out:
for x in gen:
    print(x)
```

### Equality and identity

*Equality of values* is tested with the operator `==`,
*equality of references* with the keyword `is`,
which amounts to equality of the `id`s.

```python
print(123 == 121 + 2)
a = [1,2,3]
b = a[:] # copy
print(id(a) == id(b))
print(a is b)
print(a is [1,2,3])
print(a == b)
```

*Membership of a type* we test with the method `isinstance`.

```python
print(isinstance("a", str))
print(isinstance(123, bool))
print(type(123))
print(isinstance(123, object))
```

Weaker still than `==` is a comparison of the representing strings (which is something other than a type conversion to the type `str`). Here `repr()` is supposed to deliver a string that is as unambiguous as possible, while `str()` is supposed to deliver one that is as readable as possible. Often the `repr` string can be evaluated as Python code with `eval`, which produces a copy of the object.

```python
a = dict(key1=1, key2=2)
b = {"key1":'1', "key2":'2'}
print("genuinely different dict objects have different representations:")
print("(", a, "==", b, ") =", a == b)
print("but for dict objects str() is the same as repr()")
print("(", repr(a), "==", str(a), ") = ", repr(a) == str(a))

c = eval(repr(a))
print(c)

print(str(3) == str("3"))
print(repr(3) == repr("3"))
print(repr(3), "!=", repr("3"))
```

### Control flow, f-strings and files

Loops and branches look as they do below in Python; a condition needs no brackets.

Indentation is **four spaces** – not a rule of the interpreter, but one of the style guide PEP 8, and every tool in the ecosystem assumes it. Mixing tabs and spaces is an error in Python 3.

The `for` loop is always a *for-each*: it runs over the elements of a sequence, never over a counter. If you need a counter, you create a sequence of numbers with `range(n)` and iterate over that – which is why `range` is so ubiquitous in Python. You iterate over a dictionary with `.items()`, which gives you key and value at once:

```python
for i in range(3):          # for loop over a sequence
    if i == 0:
        print(i, "is zero")
    elif i % 2 == 1:
        print(i, "is odd")
    else:
        print(i, "is even")

n = 5
while n > 0:                # while loop
    n -= 2
print("n is now", n)

for colour in ["blue", "red"]:          # for-each over a list
    print(colour)

for k, v in {"a": 1, "b": 2}.items():   # over a dict: key and value
    print(k, "->", v)
```

**f-strings** insert values into a string. After a colon you can give the formatting (width, alignment, decimal places). We use them for output from the next chapter onwards:

```python
name, share = "female", 0.7420382
print(f"share {name}: {share}")
print(f"share {name:>8s}: {share:6.2%}")   # 8 characters wide, right-aligned; percent with 2 decimal places
print(f"{2**10 = }")                       # handy when debugging
```

**Files** are opened with `open`, best inside a `with` block: it makes sure the file is closed again at the end of the block, even if an error occurs in between. An opened text file can be iterated over line by line:

```python
with open("example.txt", "w", encoding="utf-8") as file:   # "w" = write
    file.write("first line\nsecond line\n")

with open("example.txt", "r", encoding="utf-8") as file:   # "r" = read
    for number, line in enumerate(file, start=1):
        print(f"{number}: {line.strip()}")

import os
os.remove("example.txt")   # tidy up
```

### Object-oriented programming

```python
class MyClass:
    def my_method(self, methodically):
        print(methodically, self)

    def __init__(self):
        self.attribute = 123

    def __repr__(self):
        return str(self.attribute)

x = MyClass()
x.my_method("test")
print(type(x))
isinstance(x, object)
```

Polymorphism goes so far that you often do not need to know whether a variable refers to a class or to a method – if it is `Callable`, you can hang `()` after it and get something.

### Functions

A function is defined with `def`, and the return value comes from `return`. The string directly under the header line is the *docstring*, the built-in documentation that `help` displays. Parameters can have default values:

```python
def square(x, plus=0):
    """Square x and optionally add something to it."""
    return x**2 + plus

print(square(3))            # 9
print(square(3, plus=1))    # 10
help(square)
```

One detail that surprises people coming from other languages: in Python, `if`, `for` and `while` do **not** form a scope of their own. A variable created inside such a block lives on afterwards. Only functions, classes and modules have their own scope – what is assigned inside a function is not visible outside it.

Conversely, a function may *read* names from outside without taking them over. As soon as it assigns something to such a name, that name counts as local for the whole function; whoever really means the outer variable has to declare it with `global` (or, in nested functions, with `nonlocal`). Both are rarely needed and better avoided: a function that modifies variables visible from outside is hard to test.

```python
if True:
    inside = 1
print("after the if block there is", inside, "- blocks have no scope of their own")

def f():
    local = 2

f()
try:
    print(local)
except NameError as e:
    print("after the function, on the other hand, there is not:", e)
```

In a method declaration there is the option of allowing arbitrarily many positional arguments:

```python
def method(*args):
    return args

method()
print(method(1,2,3,"four"))

def new_method(must_be_there, *args):
    return args

print(new_method(1,2,3,"four"))
```

```python
try:
    new_method()
except TypeError as e:
    print(e)
```

Keyword arguments can be handled in the same way:

```python
def method(**kwargs):
    return kwargs

method()
print(method(key="value"))

try:
    method("123")
except TypeError as e:
    print(e)
```

All in all you can allow a sequence of mandatory positional arguments, a sequence of specified keyword arguments, and arbitrary further ones:

```python
def method(posarg, posarg2, *args, kwarg=None, kwarg2=True, **kwargs):
    return args, kwargs

#method() # <--- does not work

method(1, 2, 3, 4, 5, kwarg2=False, kwarg3="test")
```

To pass a whole `dict` object directly as keyword arguments, you can unpack it (and tuples in just the same way):

```python
args = (1, 2, 3, 4, 5)
kwargs = {"kwarg2": False, "kwarg3": "test"}
print(method(*args, **kwargs))

print(args)
print(*args)
```

Object attributes, too, are managed in a `dict`:

```python
class NewClass:
    def __init__(self):
        self.x = 1
        self.y = 2

new_object = NewClass()
print(new_object.__dict__)
new_object.__dict__["z"] = 3
new_object.z == 3
```
